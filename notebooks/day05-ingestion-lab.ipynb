```python
import pdfplumber
from docx import Document
```

```python
pdf_path = "../docs/benefits.pdf"

pages_text = []

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        text = page.extract_text()
        if text:
            pages_text.append(text)

benefits_text = "\n\n".join(pages_text)

print(benefits_text[:2000])
```

```python
with open("../raw_text/benefits.txt", "w", encoding="utf-8") as f:
    f.write(benefits_text)
```

```python
doc = Document("../docs/claims-process.docx")

claims_text = "\n".join(
    paragraph.text
    for paragraph in doc.paragraphs
    if paragraph.text.strip()
)

print(claims_text)
```

```python
with open("../raw_text/claims_process.txt", "w", encoding="utf-8") as f:
    f.write(claims_text)
```

### OCR Accuracy Notes

- Most printed text was extracted correctly.
- The checkbox symbols were not interpreted reliably. For example, the selected "Silver HMO" checkbox was represented as `@` rather than a checkbox.
- Some characters were misread, such as `| confirm` instead of `I confirm`.
- Some quotation-like or stray characters appeared before headings and fields.
- Names, dates, phone numbers, and email addresses were extracted correctly in this example.
- Checkbox selections and handwritten fields should be manually verified when accuracy is important.

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.cms.gov/medical-bill-rights/help/guides/health-insurance-terms"

response = requests.get(url)
response.raise_for_status()

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
main = soup.find("main")

print(main.get_text("\n", strip=True)[:3000])
```

```python
article = soup.find("article")

print(article.get_text("\n", strip=True)[:3000])
```

```python
main = soup.find("main")

# Remove navigation/header/footer elements from main
for element in main.find_all(["nav", "header", "footer"]):
    element.decompose()

web_text = main.get_text("\n", strip=True)

print(web_text[:3000])
```

```python
main = soup.find("main")

web_text = main.get_text("\n", strip=True)

start = web_text.find("Health insurance terms you should know")

if start != -1:
    web_text = web_text[start:]

print(web_text[:3000])
```

```python
import re

def clean_text(text):
    # Fix common encoding issues
    text = text.encode("utf-8", errors="ignore").decode("utf-8")

    # Replace multiple spaces/tabs with a single space
    text = re.sub(r"[ \t]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    # Remove whitespace from each line
    text = "\n".join(line.strip() for line in text.splitlines())

    return text.strip()
```

```python
benefits_text = clean_text(benefits_text)
claims_text = clean_text(claims_text)
enrollment_text = clean_text(enrollment_text)
web_text = clean_text(web_text)
```

```python
with open("../raw_text/benefits.txt", "w", encoding="utf-8") as f:
    f.write(benefits_text)

with open("../raw_text/claims_process.txt", "w", encoding="utf-8") as f:
    f.write(claims_text)

with open("../raw_text/enrollment.txt", "w", encoding="utf-8") as f:
    f.write(enrollment_text)

with open("../raw_text/provider_faq.txt", "w", encoding="utf-8") as f:
    f.write(web_text)
```
